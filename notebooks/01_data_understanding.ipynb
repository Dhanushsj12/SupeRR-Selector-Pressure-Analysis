```python
import pandas as pd
import os

DATA_PATH = "../Data Pack/"

# Load all six datasets

competitions = pd.read_csv(DATA_PATH + "competitions.csv")

teams = pd.read_csv(DATA_PATH + "teams.csv")

matches = pd.read_csv(DATA_PATH + "matches.csv")

ball_by_ball = pd.read_csv(DATA_PATH + "ball_by_ball.csv")

player_mapping = pd.read_csv(
    DATA_PATH + "competition_team_player_mapping.csv"
)

playing_eleven = pd.read_csv(
    DATA_PATH + "playing_eleven.csv"
)

print("All datasets loaded successfully!")
```

```python
datasets = {
    "Competitions": competitions,
    "Teams": teams,
    "Matches": matches,
    "Ball By Ball": ball_by_ball,
    "Player Mapping": player_mapping,
    "Playing Eleven": playing_eleven
}

for name, df in datasets.items():
    print(f"\n{name}")
    print("Rows:", df.shape[0])
    print("Columns:", df.shape[1])
```

```python

for name, df in datasets.items():

    print("\n" + "=" * 80)
    print(f"DATASET: {name.upper()}")
    print("=" * 80)

    column_info = pd.DataFrame({
        "Column": df.columns,
        "Data Type": df.dtypes.values,
        "Non Null Count": df.notna().sum().values,
        "Missing Count": df.isna().sum().values,
        "Unique Values": df.nunique().values
    })

    display(column_info)
```

```python

print("BALL BY BALL NUMERICAL SUMMARY")

display(
    ball_by_ball.describe().T
)

print("\nMATCH DATA SUMMARY")

display(
    matches.describe(include="all").T
)
```

```python

for name, df in datasets.items():

    missing_report = pd.DataFrame({
        "Missing Count": df.isnull().sum(),
        "Missing Percentage": (
            df.isnull().mean() * 100
        ).round(2)
    })

    missing_report = missing_report[
        missing_report["Missing Count"] > 0
    ].sort_values(
        by="Missing Percentage",
        ascending=False
    )

    print("\n", name.upper())

    display(missing_report)
```
